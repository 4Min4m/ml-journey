```python
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
hours_studied = np.array([1,2,3,4,5]).reshape(-1,1)
exam_scores = np.array([50,55,65,70,80])

print("Hours Studied: ",hours_studied)
print("Exam Scores: ",exam_scores)
print("---")
print("With 3 hours studing, score is 65")
```

```python
model = LinearRegression()
model.fit(hours_studied,exam_scores)


print("Model Trained!")
print("Coefficient: ",model.coef_[0])
print("Intercept: ",model.intercept_)
```

```python
new_hours = np.array([[6]])
predicted_score = model.predict(new_hours)

print(f"👨‍🎓 Predict: {new_hours[0][0]} hours study, score is : {predicted_score[0]:.1f}")

print(f"👩‍🎓 predict: with 2 and half hours study, score is {model.predict([[2.5]])[0]:.1f}")
```

```python
import matplotlib.pyplot as plt

plt.scatter(hours_studied,exam_scores)
plt.plot(hours_studied,model.predict(hours_studied))
plt.xlabel("Hours Studied")
plt.ylabel("Exam Scores")
plt.title("Hours Studied vs Exam Scores")
plt.grid(True)
plt.show()
```
